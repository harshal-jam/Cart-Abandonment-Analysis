```python
df5=pd.read_csv ("C:/Users/hp/Documents/toys/data/clean/website_sessions.csv",parse_dates=["created_at"]);
```

```python
df5.isna().sum()
```

```python
df5.isna().mean()
```

```python
df5["utm_campaign"].mode()
```

```python
mode_value = df5["utm_campaign"].mode()[0]
df5["utm_campaign"] = df5["utm_campaign"].fillna(mode_value)
```

```python
df5['utm_campaign'].isna().sum()
```

```python
category = ["utm_source","utm_content","http_referer"]
for col in category:
    mode_values = df5[col].mode()[0]
    df5[col]=df5[col].fillna(mode_values)
```

```python
df5.isna().sum()
```

```python
df5.to_csv("C:/Users/hp/Documents/toys/data/clean/website_sessions.csv",index=False);
```
